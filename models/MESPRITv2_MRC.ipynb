```python
# ===========================
# MORTALITY SCORE - DIRECT CALCULATION (NO ML MODEL NEEDED!)
# Based on clinical vitals only
# ===========================

def calculate_mortality_score_direct(age, gender, diagnosis, body_temp, heart_rate,
                                     resp_rate, oxygen_sat, systolic_bp, diastolic_bp,
                                     family_history='No', smoking='Never', alcohol='No'):
#APACHE II approach
    score = 0

    # =============================
    # AGE FACTOR (Max 20 points)
    # =============================
    if age >= 85:
        score += 20
    elif age >= 75:
        score += 16
    elif age >= 65:
        score += 12
    elif age >= 55:
        score += 8
    elif age >= 45:
        score += 4

    # =============================
    # OXYGEN SATURATION (Max 30 points) - MOST IMPORTANT
    # =============================
    if oxygen_sat < 75:
        score += 30
    elif oxygen_sat < 80:
        score += 28
    elif oxygen_sat < 85:
        score += 25
    elif oxygen_sat < 88:
        score += 22
    elif oxygen_sat < 90:
        score += 18
    elif oxygen_sat < 92:
        score += 14
    elif oxygen_sat < 94:
        score += 10
    elif oxygen_sat < 95:
        score += 6
    elif oxygen_sat < 96:
        score += 3

    # =============================
    # BLOOD PRESSURE (Max 20 points)
    # =============================
    if systolic_bp < 70:
        score += 20
    elif systolic_bp < 80:
        score += 18
    elif systolic_bp < 90:
        score += 15
    elif systolic_bp < 100:
        score += 10
    elif systolic_bp > 200:
        score += 20
    elif systolic_bp > 180:
        score += 15
    elif systolic_bp > 160:
        score += 8
    elif systolic_bp > 140:
        score += 4

    # =============================
    # HEART RATE (Max 15 points)
    # =============================
    if heart_rate < 40 or heart_rate > 150:
        score += 15
    elif heart_rate < 50 or heart_rate > 140:
        score += 12
    elif heart_rate < 55 or heart_rate > 130:
        score += 9
    elif heart_rate < 60 or heart_rate > 120:
        score += 7
    elif heart_rate > 110:
        score += 5
    elif heart_rate > 100:
        score += 3

    # =============================
    # RESPIRATORY RATE (Max 10 points)
    # =============================
    if resp_rate > 40:
        score += 10
    elif resp_rate > 35:
        score += 9
    elif resp_rate > 30:
        score += 7
    elif resp_rate > 25:
        score += 5
    elif resp_rate > 22:
        score += 3
    elif resp_rate > 20:
        score += 2
    elif resp_rate < 8:
        score += 10
    elif resp_rate < 10:
        score += 7

    # =============================
    # BODY TEMPERATURE (Max 10 points)
    # =============================
    if body_temp >= 42 or body_temp < 35:
        score += 10
    elif body_temp >= 41 or body_temp < 35.5:
        score += 8
    elif body_temp >= 40 or body_temp < 36:
        score += 6
    elif body_temp >= 39.5:
        score += 4
    elif body_temp >= 39:
        score += 3
    elif body_temp >= 38.5:
        score += 2

    # =============================
    # DIAGNOSIS (Max 15 points)
    # =============================
    critical_diseases = ['Pneumonia', 'COVID-19', 'Tuberculosis', 'Chronic Kidney Disease']
    serious_diseases = ['Hepatitis', 'Asthma', 'Bronchitis', 'Influenza', 'Dengue', 'Malaria']
    chronic_diseases = ['Hypertension', 'Diabetes Mellitus']

    if diagnosis in critical_diseases:
        score += 15
    elif diagnosis in serious_diseases:
        score += 10
    elif diagnosis in chronic_diseases:
        score += 5
    elif diagnosis == 'Healthy':
        score += 0
    else:
        score += 7

    # =============================
    # RISK FACTORS (Max 5 points)
    # =============================
    if smoking in ['Yes', 'Former']:
        score += 2
    if family_history == 'Yes':
        score += 2
    if alcohol == 'Yes':
        score += 1

    # Cap at 100
    score = min(score, 100)

    # Determine category
    if score >= 70:
        category = 'High Risk'
        color = '🔴'
    elif score >= 40:
        category = 'Medium Risk'
        color = '🟡'
    else:
        category = 'Low Risk'
        color = '🟢'

    # Display results
    print(f"\n{'='*75}")
    print(f" MORTALITY SCORE CALCULATION")
    print(f"{'='*75}")

    print(f"\n PATIENT INFO:")
    print(f"   Age: {age}y | Gender: {gender} | Diagnosis: {diagnosis}")

    print(f"\n VITAL SIGNS:")
    print(f"   Temperature: {body_temp}°C")
    print(f"   Heart Rate: {heart_rate} bpm")
    print(f"   Respiratory Rate: {resp_rate} breaths/min")
    print(f"   Oxygen Saturation: {oxygen_sat}%")
    print(f"   Blood Pressure: {systolic_bp}/{diastolic_bp} mmHg")

    print(f"\n MORTALITY SCORE: {score:.0f} / 100")

    # Visual bar
    filled = int(score / 2)
    empty = 50 - filled
    bar = '█' * filled + '░' * empty
    print(f"   [{bar}] {score:.0f}%")

    print(f"\n{color} RISK CATEGORY: {category}")

    if score >= 70:
        print(f"CRITICAL - Immediate intervention required!")
    elif score >= 40:
        print(f"CAUTION - Close monitoring needed")
    else:
        print(f" STABLE - Routine care appropriate")

    print(f"\n SCORE BREAKDOWN:")
    print(f"   Age Factor: Contributing to risk")
    print(f"   Oxygen Level: {'CRITICAL' if oxygen_sat < 90 else 'CONCERNING' if oxygen_sat < 94 else 'NORMAL'}")
    print(f"   Blood Pressure: {'ABNORMAL' if systolic_bp < 90 or systolic_bp > 140 else 'NORMAL'}")
    print(f"   Heart Rate: {'ABNORMAL' if heart_rate < 60 or heart_rate > 100 else 'NORMAL'}")
    print(f"   Disease Severity: {diagnosis}")

    print(f"{'='*75}\n")

    return score, category



```
